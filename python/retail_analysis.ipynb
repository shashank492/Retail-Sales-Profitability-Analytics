```python
columns = ["Quantity", "Sales", "Cost", "Profit"]

for col in columns:
    Q1 = sales_clean[col].quantile(0.25)
    Q3 = sales_clean[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = (
        (sales_clean[col] < lower) |
        (sales_clean[col] > upper)
    ).sum()

    print(f"{col}: {outliers} outliers")
```
